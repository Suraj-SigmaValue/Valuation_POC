```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv

load_dotenv()

GOOGLE_PLACES_API_KEY = os.getenv("GOOGLE_PLACES_API_KEY")

if not GOOGLE_PLACES_API_KEY:
    raise ValueError("Google Places API key not found in .env")

url = "https://maps.googleapis.com/maps/api/place/textsearch/json"

params = {
    "query": "Riverdale Residences Pune",
    "key": GOOGLE_PLACES_API_KEY
}

response = requests.get(url, params=params, timeout=30).json()

if response.get("status") != "OK":
    raise RuntimeError(
        f"Google Places API error: {response.get('status')} - "
        f"{response.get('error_message', '')}"
    )

# Extract every matching result returned by the API
projects = []

for place in response.get("results", []):
    projects.append({
        "project_name": place.get("name"),
        "address": place.get("formatted_address"),
        "place_id": place.get("place_id"),
        "primary_types": place.get("types", [])
    })

df = pd.DataFrame(projects)

display(df)
```
